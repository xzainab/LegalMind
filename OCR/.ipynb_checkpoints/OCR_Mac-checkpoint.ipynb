```python
import os
from pdf2image import convert_from_path
from ocrmac import ocrmac

# Define paths
pdf_path = "1_د_03_1.pdf"
output_txt_path = "t3.txt"

print("Converting PDF pages to images... (This may take a moment)")
pages = convert_from_path(pdf_path, dpi=300)

print("Starting Apple Vision OCR...")
with open(output_txt_path, "w", encoding="utf-8") as f:
    for i, page in enumerate(pages):
        print(f"Processing page {i+1}/{len(pages)}...")
        
        # Save temp image because ocrmac reads from paths
        temp_img_path = f"temp_page_{i}.png"
        page.save(temp_img_path, "PNG")
        
        # Run Apple's native text recognition (configured for Arabic)
        # recognition_level=0 means "Accurate" mode
        annotations = ocrmac.OCR(temp_img_path, language_preference=['ar-SA'], recognition_level='accurate').recognize()
        
        # Extract text strings from the annotations structure
        page_text = "\n".join([text_block[0] for text_block in annotations])
        
        # Clean up temporary page image
        os.remove(temp_img_path)
        
        # Write to file
        f.write(f"--- Page {i+1} ---\n")
        f.write(page_text)
        f.write("\n\n")

print(f"Success! Highly accurate text saved to {output_txt_path}")
```
