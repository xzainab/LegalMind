```python
import pypdf
from bidi.algorithm import get_display
import arabic_reshaper

# 1. Open the PDF file
reader = pypdf.PdfReader("17-42-Prof.-Mashari-K.-Al-Aifan-Dr.-Fares-M.-Al-Mutairi.pdf")

# 2. Loop through each page and extract text
for page in reader.pages:
    text = page.extract_text()
    
    if text.strip():  # Check if the page is not empty
        # 3. Reshape and fix the direction of the Arabic text
        reshaped_text = arabic_reshaper.reshape(text)
        bidi_text = get_display(reshaped_text)
        
        print(bidi_text)
```
