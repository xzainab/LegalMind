```python
import re
from pathlib import Path

base_path = Path("./CBB_Documents")

# Complete pattern list ordered from most specific to least specific
replacements = [
    # 4-character / 3-character codes
    (r"أ\.?\s*ج\.?\s*م\.?", "أمر جنائي مستعجل مؤقت"),
    (r"ا\.?\s*ح\.?\s*م\.?", "أمر حظر سفر أو حجز مال"),
    
    # 2-character codes
    (r"د\.?\s*ت\.?", "دعوى تجارية"),
    (r"م\.?\s*ت\.?", "منازعة تنفيذ"),
    (r"ط\.?\s*ح\.?", "طلب حبس"),
    (r"ط\.?\s*ش\.?", "طلب شرعي أو طعن شرعي"),
    
    # Single-letter codes (guarded with regex boundaries so they don't match inside full words)
    (r"(?<![أ-يa-zA-Z0-9])د(?![أ-يa-zA-Z0-9])", "دعوى موضوعية"),
    (r"(?<![أ-يa-zA-Z0-9])ن(?![أ-يa-zA-Z0-9])", "طلب تنفيذ"),
    (r"(?<![أ-يa-zA-Z0-9])ج(?![أ-يa-zA-Z0-9])", "قضية جنائية"),
    (r"(?<![أ-يa-zA-Z0-9])ح(?![أ-يa-zA-Z0-9])", "حجز أو دعوى حيازة"),
]

for file_path in base_path.rglob("*.txt"):
    # Skip Jupyter internal checkpoint folders
    if ".ipynb_checkpoints" in file_path.parts:
        continue

    stem = file_path.stem
    new_stem = stem

    # Apply replacement rules
    for pattern, replacement in replacements:
        new_stem = re.sub(pattern, replacement, new_stem)

    # Clean up artifacts (dots around underscores, multiple dots/spaces)
    new_stem = re.sub(r"\._", "_", new_stem)
    new_stem = re.sub(r"_\.", "_", new_stem)
    new_stem = re.sub(r"\s+", " ", new_stem)
    new_stem = re.sub(r"\.+", ".", new_stem).strip(". ")

    if stem != new_stem:
        new_file_path = file_path.with_name(f"{new_stem}{file_path.suffix}")
        file_path.rename(new_file_path)
        print(f"Renamed: {file_path.name} -> {new_file_path.name}")

print("\nAll files successfully updated!")
```
